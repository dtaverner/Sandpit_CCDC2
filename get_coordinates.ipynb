```python
# This code takes the list of identifiers (text file) with no associated SMILES string, iterates through this list and checks if the entry has 3D coordinates. 
# The 3D coordinate check is done by accessing the crystal and molecule from the identifier (refcode). 3D coordinates are tested by attempting to the atom at position 0 in the molecular model.
# Successful access indicates that the entry contains at least one atom with defined coordinates. Identifiers for which atomic coordinates are present are written to an output CSV file. 

from ccdc import io
import csv
from tqdm import tqdm

reader = io.EntryReader("CSD")

with open(r"C:\projectSandpit\empty_smiles_identifiers.txt") as f, \
     open("has_coordinates_only.csv", "w", newline="") as out:

    writer = csv.writer(out)
    writer.writerow(["refcode"])

    for line in tqdm(f, desc="Checking CSD entries"):
        refcode = line.strip()
        if not refcode:
            continue

        try:
            crystal = reader.crystal(refcode)
            molecule = crystal.molecule

            # Explicit coordinate test
            atom = molecule.atoms[0]

            writer.writerow([refcode])

        except Exception:
            pass
```
